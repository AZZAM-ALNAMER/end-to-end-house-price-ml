### Subsection 3.2

## Using the IQR method to numerically identify outliers and reporting their count per variable

###Steps Summary :

First, we selected all numerical columns from the dataset.

Second, we calculated the IQR (Interquartile Range) for each one to define the normal range.

Third, we counted the number of values that fall below or above the IQR boundaries — these are considered outliers.

Fourth, we saved the counts in a dictionary and displayed the top variables with their outliers.

This helps us easily identify which features have extreme values that may need special handling.

```python
# first Selecting numerical columns
num_cols = df.select_dtypes(include=['float64', 'int64']).columns

# second Creating a dictionary to store outlier counts each numerical coulmn for
outlier_summary = {}

# Third Looping through each  numerical columns and calculate IQR-based outliers
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    outlier_summary[col] = len(outliers)

# Fourth Converting the python code output in to dataframe so we can see the result clearly
outlier_report = pd.DataFrame.from_dict(outlier_summary, orient='index', columns=['Number of Outliers'])
print(outlier_report)
```

### Subsection 3.3 :

## Performing outlier detection using the Z-score method and compare it with IQR results.

### Steps summary:

First, we selected the same numerical columns we used for IQR.

Second, we calculated the Z-score for each value  this tells how many standard deviations a value is far from the mean.

Third, we set a threshold of 3 beause it is common measure value for many coders , meaning values beyond ±3 standard deviations are considered outliers.

Fourth, we counted how many extreme values exist for each column.

lastly, we compared the results with the IQR method to see which variables are showing high outlier counts

```python
from scipy import stats  # we import scipy since we need to work wih advance statical math like z-score which can not handle by nunmpy alone
import numpy as np

# first Selecting numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# second calacuating Z-scores
z_scores = np.abs(stats.zscore(df[num_cols]))

# Third Define a threshold
threshold = 3

# Fourth Count outliers per column
z_outlier_counts = (z_scores > threshold).sum(axis=0)

# lastly Create a comparison  Z-score DataFrame with IQR pervious  results
z_outlier_df = pd.DataFrame({
    'IQR_Outliers': outlier_report['Number of Outliers'],
    'Zscore_Outliers': z_outlier_counts
}).sort_values(by='Zscore_Outliers', ascending=False)

display(z_outlier_df.head(10))
```

##Section 6 :  Data Transformations and Normalization



### Subsetion 6.1

###Encoding Categorical Variables (Label Encoding + One-Hot Encoding)


###Steps Summary:

**First** we used the `label-encdoingd` Normalization methhod for the columns they have pattern and meangiful order in their values for example `HouseAge` has a natural order: New , Normal , Old ,
Since these categories follow a clear ranking, turning them to numbers like 0 → 1 → 2 makes sense , Note( in the output we will get True or False and it is normal sinnce most of ML models directly convert them to 1-0 values )

**Second** :  we used in the other` hand One-Hot Encoding` for those features do not have order like `HouseStyle` where Where each column is 0 or 1. This avoids giving fake meaning or ranking to categories.



```python
from sklearn.preprocessing import LabelEncoder

# First we use the labale encoding for the ordinal features
label_cols = [
    'ExterQual', 'ExterCond','HouseAge', 'BsmtQual', 'BsmtCond', 'BsmtExposure',
    'BsmtFinType1', 'BsmtFinType2', 'HeatingQC', 'KitchenQual',
    'Functional', 'FireplaceQu', 'GarageQual', 'GarageCond',
    'GarageFinish', 'PavedDrive', 'CentralAir',
]

le = LabelEncoder()

for col in label_cols:
    # Check if the column exists in df before trying to encode it
    if col in df.columns:
        df[col + "_LE"] = le.fit_transform(df[col].astype(str))
    else:
        print(f"Warning: Column '{col}' not found for Label Encoding, skipping.")

# Second we use the One-Hot Encoding for the nominal columns

onehot_cols_static = [
    'MSZoning', 'Street', 'LotShape', 'LandContour', 'Utilities',
    'LotConfig', 'LandSlope', 'Neighborhood', 'Condition1', 'Condition2',
    'BldgType', 'HouseStyle', 'RoofStyle', 'RoofMatl', 'Exterior1st',
    'Exterior2nd', 'MasVnrType', 'Foundation', 'Heating', 'Electrical',
    'GarageType', 'SaleType', 'SaleCondition'
]

# Filter onehot_cols_static to include only columns that actually exist in df
# This is crucial for avoiding the KeyError if some columns were dropped previously
onehot_cols_to_process = [col for col in onehot_cols_static if col in df.columns]

if not onehot_cols_to_process:
    print("No suitable object columns from the specified list were found for One-Hot Encoding.")
else:
    df = pd.get_dummies(df, columns=onehot_cols_to_process, drop_first=True)

# Lastly displaying the head of the dataset
df.head()
```

##Section2 :Baseline Models

### 2.1 Linear Regression Model

we will train a standard Linear Regression model and evaluate its performance on both training and testing datasets using multiple evaluation metrics.

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import numpy as np

# Impute any remaining NaNs in X_train with the mean of its columns
X_train = X_train.fillna(X_train.mean())

# Impute any remaining NaNs in X_test with the mean of the corresponding columns from X_train
X_test = X_test.fillna(X_train.mean())


# fiest we  Train Linear Regression Model

linear_model = LinearRegression()
linear_model.fit(X_train, y_train)


#  second Making Predictions

y_train_pred = linear_model.predict(X_train)
y_test_pred = linear_model.predict(X_test)


# Lastly Calculate Evaluation Metrics seperartly


# R² Score
r2_train = r2_score(y_train, y_train_pred)
r2_test = r2_score(y_test, y_test_pred)

# RMSE
rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))

# MAE
mae_train = mean_absolute_error(y_train, y_train_pred)
mae_test = mean_absolute_error(y_test, y_test_pred)

# MAPE
mape_train = np.mean(np.abs((y_train - y_train_pred) / y_train)) * 100
mape_test = np.mean(np.abs((y_test - y_test_pred) / y_test)) * 100

# showing the results
print("Training Metrics:")
print(f"R² Score: {r2_train:.4f}")
print(f"RMSE: {rmse_train:.4f}")
print(f"MAE: {mae_train:.4f}")
print(f"MAPE: {mape_train:.2f}%")

print("\nTesting Metrics:")
print(f"R² Score: {r2_test:.4f}")
print(f"RMSE: {rmse_test:.4f}")
print(f"MAE: {mae_test:.4f}")
print(f"MAPE: {mape_test:.2f}%")
```

 we Printed evaluation metrics for Ridge Regression and a coefficient comparison table showing shrinkage effects as shown below

###4.2 Tuned Model Evaluation

In this step, we use GridSearchCV to find the best combinations of hyperparameters for our XGBoost model because GridSearchCV automatically tests many combinations of hyperparameter values and selects the combination that gives the best performance based on cross-validation We use 5-fold cross-validation to ensure the selected parameters generalize well to unseen data.

` cell output`

The result as it shows below that cross-validation successfully evaluated multiple hyperparameter combinations and selected a well-balanced model that controls complexity max_depth=3 while achieving strong performance n_estimators=200, learning_rate=0.1)

```python
from sklearn.model_selection import GridSearchCV
import xgboost as xg

# First Define the parameter grid with 3+ hyperparameters
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1, 0.2]
}

# Second Initialize GridSearchCV with 5-fold cross-validation
grid_search = GridSearchCV(
    estimator=xg.XGBRegressor(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring='neg_mean_squared_error',
    n_jobs=-1,
    verbose=1
)

# Third Fit the grid search to the training data
grid_search.fit(X_train, y_train)

# Fourth Report the best parameters
best_params = grid_search.best_params_
print("Best Parameters Found:")
print(best_params)

# Lastly Store the best model for final evaluation
best_xgb_model = grid_search.best_estimator_
```

`cell output`
